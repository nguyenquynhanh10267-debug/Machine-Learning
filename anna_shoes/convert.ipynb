```python
import pandas as pd
from deep_translator import GoogleTranslator
from tqdm import tqdm
import re
```

```python
INPUT_FILE = r'C:\Users\AWESOME\Machine-Learning\anna_shoes\anta_shoes_raw.csv'
OUTPUT_FILE = r'C:\Users\AWESOME\Machine-Learning\anna_shoes\anta_shoes_trans.csv'

translator = GoogleTranslator(source="zh-CN", target="vi")

# Regex nhận biết có ký tự tiếng Trung
pattern = re.compile(r'[\u4e00-\u9fff]')
```

```python
print("Đang đọc file...")

df = pd.read_csv(INPUT_FILE)

print(f"Số dòng: {len(df):,}")
print(f"Số cột : {len(df.columns)}")
```

```python
print("\nĐang dịch tên cột...")

column_mapping = {}

for col in tqdm(df.columns):

    if isinstance(col, str) and pattern.search(col):
        try:
            column_mapping[col] = translator.translate(col)
        except:
            column_mapping[col] = col
    else:
        column_mapping[col] = col

df.rename(columns=column_mapping, inplace=True)

print("Đã dịch xong tên cột.")
```

```python
df.columns
```

```python
columns_translate = [
    '品牌', '类目', '性别','省份','城市', '城市等级','网络类型','来源渠道',
    '入口关键词','行为序列'
]

# đổi sang tên sau khi rename

columns_translate = [
    column_mapping.get(col, col)
    for col in columns_translate
]
```

```python
cache = {}

for col in columns_translate:

    if col not in df.columns:
        print(f"Bỏ qua cột {col}")
        continue

    print(f"\nĐang dịch cột: {col}")

    unique_values = df[col].dropna().unique()

    print(f"Có {len(unique_values)} giá trị unique")

    mapping = {}

    for value in tqdm(unique_values):

        if not isinstance(value, str):
            mapping[value] = value
            continue

        if not pattern.search(value):
            mapping[value] = value
            continue

        if value in cache:
            mapping[value] = cache[value]
            continue

        try:
            translated = translator.translate(value)
            mapping[value] = translated
            cache[value] = translated
        except:
            mapping[value] = value

    df[col] = df[col].map(mapping)
```

```python
print("\nĐang lưu file...")

df.to_csv(OUTPUT_FILE, index=False)

print("Hoàn thành!")
print("File lưu tại:", OUTPUT_FILE)
```
